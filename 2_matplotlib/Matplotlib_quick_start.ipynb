## Demos

### Import

```python
%matplotlib inline
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-darkgrid")
import numpy as np
```

### First learn what are the available styles

```python
plt.style.available
```

```python
fig = plt.figure()
ax = plt.axes()
```

### Plotting the Sin(x) function with `ax.plot()` (or just use `plt.plot()` where the axes are created in the background)

```python
fig = plt.figure()
ax = plt.axes()

x = np.linspace(0, 10, 100)
ax.plot(x, np.sin(x));
```

### Styles can get very detailed

```python
y = np.cos(x)

plt.plot(x, y, color='green', alpha=0.5, linewidth=3, linestyle='--', marker='o', markersize=10, markerfacecolor='blue', markeredgewidth=3, markeredgecolor='red')  
```

### Plot two functions on the same axes

```python
plt.plot(x, np.cos(x))
plt.plot(x, np.sin(x))
```

### Adjusting the line color styles

```python
plt.plot(x, np.sin(x - 0), color="black")  # specify color by name
plt.plot(x, np.sin(x - 1), color="g")  # short color code (rgbcmyk)
plt.plot(x, np.sin(x - 2), color="0.75")  # grayscale between 0 and 1
plt.plot(x, np.sin(x - 3), color="#FFDD44")  # hex code (RRGGBB, 00 to FF)
plt.plot(x, np.sin(x - 4), color=(1.0, 0.2, 0.3))  # RGB tuple, values 0 to 1
plt.plot(x, np.sin(x - 5), color="chartreuse");  # HTML color names supported
```

### Adjusting line styles

```python
plt.plot(x, x + 0, linestyle="solid")
plt.plot(x, x + 1, linestyle="dashed")
plt.plot(x, x + 2, linestyle="dashdot")
plt.plot(x, x + 3, linestyle="dotted")
# For short, you can use the following codes:
plt.plot(x, x + 4, linestyle="-")  # solid
plt.plot(x, x + 5, linestyle="--")  # dashed
plt.plot(x, x + 6, linestyle="-.")  # dashdot
plt.plot(x, x + 7, linestyle=":");  # dotted
```

### More compact style: color and line styles combined

```python
plt.plot(x, x + 0, "-g")  # solid green
plt.plot(x, x + 1, "--c")  # dashed cyan
plt.plot(x, x + 2, "-.k")  # dashdot black
plt.plot(x, x + 3, ":r");  # dotted red
```

### Adjusting Axis Ratio

```python
# tight x and y axes, no extra whitespace
plt.plot(x, np.sin(x))
plt.axis("tight");
```

```python
# equal x and y axes, same scaling
plt.plot(x, np.sin(x))
plt.axis("equal");
```

```python
# turn off axes lines and labels
plt.plot(x, np.sin(x))
plt.axis("off");
```

```python
# square axes, equal scaling, no extra whitespace
plt.plot(x, np.sin(x))
plt.axis("square");
```

```python
# image axes
plt.plot(x, np.sin(x))
plt.axis("image");
```

### Adding x and y axes labels and title

```python
plt.plot(x, np.sin(x))
plt.title("A Sine Curve")
plt.xlabel("x")
plt.ylabel("sin(x)");
```

### Adding Ledgend

```python
plt.plot(x, np.sin(x), "-g", label="sin(x)")
plt.plot(x, np.cos(x), ":b", label="cos(x)")
plt.axis("equal")

plt.legend();
```
